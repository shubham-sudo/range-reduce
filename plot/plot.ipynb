```python
import matplotlib.pyplot as plt
import pandas as pd

f1 = "/Users/shubham/SSD Lab/range-reduce/.vstats/experiments-testing-I100000-U100000-Q1000-D0-S1000-Y0.1-R0-y0-E128-B32-P16-T4/range-reduce/stats.log"
f2 = "/Users/shubham/SSD Lab/range-reduce/.vstats/experiments-testing-I100000-U100000-Q1000-D0-S1000-Y0.1-R0-y0-E128-B32-P16-T4/rocksdb/stats.log"
```

```python
rr_get_time = list()
rocksdb_get_time = list()

with open(f1, "r") as file1:
    for line in file1:
        if "GetTime:" in line:
            parts = line.split()
            rr_get_time.append(float(parts[1]))

with open(f2, "r") as file2:
    for line in file2:
        if "GetTime:" in line:
            parts = line.split()
            rocksdb_get_time.append(float(parts[1]))

if (len(rr_get_time) > 0 and len(rocksdb_get_time) > 0):
    plt.plot(rr_get_time, label='RangeReduce')
    plt.plot(rocksdb_get_time, label='RocksDB', alpha=0.7)
    plt.xlabel('point query number')
    plt.ylabel('latency (ns)')
    plt.ylim(bottom=0)
    plt.legend(frameon=False)
    plt.show()
    print("average RangeReduce get time:", sum(rr_get_time)/len(rr_get_time))
    print("average RocksDB get time:", sum(rocksdb_get_time)/len(rocksdb_get_time))
    print("good/bad: ", (sum(rocksdb_get_time) - sum(rr_get_time))/sum(rocksdb_get_time))
    print("good/bad: ", (sum(rocksdb_get_time)/len(rocksdb_get_time) - sum(rr_get_time)/len(rr_get_time))/sum(rocksdb_get_time)/len(rocksdb_get_time))
```
